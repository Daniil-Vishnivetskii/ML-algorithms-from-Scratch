## Linear Regression with L2 Regularization (Ridge)

The loss function is:
$$L_{ridge}=||\mathbf X\mathbf w-\mathbf y||_2^2+\alpha||\mathbf w||_2^2$$


Implementation using the closed-form solution:

$$\mathbf{w} = (\mathbf{X}^T\mathbf{X} + \alpha\mathbf{I})^{-1}\mathbf{X}^T\mathbf{y}$$

where $\alpha$ is the regularization parameter.

```python
class RidgeRegression:
    def __init__(self, alpha=1.0):
        self.alpha = alpha
        self.weights = None
    
    def fit(self, X, y):
        sz = X.shape[1]
        self.weights = np.linalg.inv(X.T @ X + self.alpha * np.diag(np.ones(sz))) @ X.T @ y
        
    def predict(self, X):
        return X @ self.weights
```

## Linear Regression with L1 Regularization (Lasso)

The loss function is:
$$L_{lasso} = \frac{1}{n}||\mathbf{Xw} - \mathbf{y}||_2^2 + \alpha||\mathbf{w}||_1$$


Implementation using subgradient descent:

$$\mathbf{w}_{t+1} = \mathbf{w}_{t} - \eta \left(\frac{2}{n} \mathbf X^T (\mathbf X\mathbf w-\mathbf y) + \alpha\, sign(\mathbf w) \right)$$

where $\alpha$ is the regularization parameter, $\eta$ is the learning rate.

```python
class LassoRegression:
    def __init__(self, alpha=1.0, learning_rate=0.01, n_iterations=1000):
        self.alpha = alpha
        self.learning_rate = learning_rate
        self.n_iterations = n_iterations
        self.weights = None
    
    def fit(self, X, y):
        sz = y.size
        self.weights = np.zeros(X.shape[1])
        for _ in range(self.n_iterations):
            self.weights -= self.learning_rate * (
                2 / sz * X.T @ (X @ self.weights - y) 
                + self.alpha * np.sign(self.weights) 
                )
    
    def predict(self, X):
        return X @ self.weights
```

## Linear Regression with Elastic Net Regularization

Elastic Net combines L1 and L2 regularization:

$$L_{elastic}=
\frac1n||Xw-y||^2
+
\alpha\,
\left(
\rho ||w||_1
+
\frac{1-\rho}{2}
||w||_2^2
\right)$$

Implementation using subgradient descent:

$$\mathbf w_{t+1}
=
\mathbf w_t
-
\eta
\left(
\frac{2}{n}
\mathbf X^T
(
\mathbf X\mathbf w-\mathbf y
)
+
\alpha\,
\Big(
(1-\rho)\mathbf w
+
\rho\,sign(\mathbf w)
\Big)
\right)$$

where:
- $\eta$ is the learning rate
- $\alpha$ is the regularization strength
- $\rho$ is the mixing parameter (0 ≤ ρ ≤ 1)
  - ρ = 1: Pure Lasso (L1)
  - ρ = 0: Pure Ridge (L2)
  - 0 < ρ < 1: Mix of both

```python
class ElasticNetRegression:
    def __init__(self, alpha=1.0, rho=0.5, learning_rate=0.01, n_iterations=1000):
        self.alpha = alpha
        self.rho = rho
        self.learning_rate = learning_rate
        self.n_iterations = n_iterations
        self.weights = None
    
    def fit(self, X, y):
        sz = y.size
        self.weights = np.zeros(X.shape[1])
        for _ in range(self.n_iterations):
            self.weights -= self.learning_rate * (
                2 / sz * X.T @ (X @ self.weights - y) 
                + self.alpha * (1 - self.rho) * self.weights 
                + self.alpha * self.rho * np.sign(self.weights) 
                ) 
    
    def predict(self, X):
        return X @ self.weights
```
